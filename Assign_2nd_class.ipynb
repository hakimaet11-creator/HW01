```python
# 5. Create a list of 5 numbers, add a new number, remove one, and then sort the list.
numbers = [1, 2, 3, 4, 5]
numbers.append(6)
numbers.remove(3)
print(numbers)
numbers.sort()
print(numbers)
numbers.reverse()
print(numbers)
```

```python
# 8. Create two sets of colors and find their union and intersection.
primary_colors = {"red", "yellow", "blue"}
rainbow_colors = {"red", "orange", "yellow", "green", "blue", "indigo", "violet"}
all_colors = primary_colors | rainbow_colors
print(all_colors)
common_colors = primary_colors & rainbow_colors
print(common_colors)
```
